# Handling Outliers

```python
df_new = df
df_new.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.displot(df_new['Price'], kde=True)
plt.show()
# Data is not normally distributed
```

Z Score Method

```python
from scipy import stats

df_new["Z_Score"] = abs(stats.zscore(df_new["Price"]))
df_new = df_new[df_new["Z_Score"] < 3]
df_new.drop(columns=["Z_Score"], inplace=True)
df_new.info()
```

```python
df_new['Price'].describe()
```

```python
sns.boxplot(df_new['Price'])
plt.show()
```

```python
iqr = df_new['Price'].quantile(0.75) - df_new['Price'].quantile(0.25)
upper_limit = df_new['Price'].quantile(0.75) + 1.5*iqr
lower_limit = df_new['Price'].quantile(0.25) - 1.5*iqr
print('Upper Limit:',upper_limit)
print('Lower Limit:',lower_limit)

df_new = df_new[(df_new["Price"] >= lower_limit) & (df_new["Price"] <= upper_limit)]
df_new.info()
# No outliers found in the column
```
